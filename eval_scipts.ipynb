```python
import pickle
import os

# 替换为您实际的文件路径
def ReadLogger(path, steps=1e6):
    with open(path, "rb") as f:
        logger = pickle.load(f)
    return logger.tracker['reward'][:int(steps)]
```

```python
reward_k_top = ReadLogger("results/qvpo_HalfCheetah-v3/25_12_26_08_13_10")
reward_baseline = ReadLogger("results/qvpo_HalfCheetah-v3/25_12_28_22_49_28")
reward_warmup = ReadLogger("results/qvpo_HalfCheetah-v3/25_12_22_12_54_55")
reward_3m_training = ReadLogger("results/qvpo_HalfCheetah-v3/25_12_16_22_14_48", steps=3e6)
reward_ratio_005 = ReadLogger("results/qvpo_HalfCheetah-v3/25_12_21_02_07_39")
reward_ratio_015 = ReadLogger("results/qvpo_HalfCheetah-v3/25_12_21_16_30_39")
reward_test = ReadLogger("results/qvpo_HalfCheetah-v3/25_12_25_23_04_25")
reward_adaptive = ReadLogger("results/qvpo_HalfCheetah-v3/25_12_16_22_14_48")
```

```python
import matplotlib.pyplot as plt
plt.plot(reward_test, label='Baseline 1', alpha=0.7, linewidth=1)
plt.plot(reward_baseline, label='Baseline 2', alpha=0.7, linewidth=1)
plt.plot(reward_adaptive, label='Adaptive noise scheduling', alpha=0.7, linewidth=1)
plt.xlabel('Steps')
plt.ylabel('Reward')
plt.title('Reward Comparison on baseline vs Adaptive noise scheduling')
plt.legend()
plt.grid()
```
